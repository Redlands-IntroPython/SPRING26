```python
# Get a random child/pet/relation... can I get a random one of these?
# What would I need to do?
# Choose a dictionary key randomly

for name in my_people.keys():
  nicknames = my_people[name]['Nicknames']
  num_nicknames = len(nicknames)
  num1 = np.random.randint(num_nicknames)

  relations = list(my_people[name]['Family'].keys())
  num_relations = len(relations)
  num2 = np.random.randint(num_relations)
  random_relation = relations[num2]

  message = f"Hey {nicknames[num1]} I hear that your....{random_relation}"
  print(message)
```

```python
# Now I need to get the relationship of that random relation

for name in my_people.keys():
  nicknames = my_people[name]['Nicknames']
  num_nicknames = len(nicknames)
  num1 = np.random.randint(num_nicknames)

  relations = list(my_people[name]['Family'].keys())
  num_relations = len(relations)
  num2 = np.random.randint(num_relations)
  random_relation = relations[num2]

  random_relationship = my_people[name]['Family'][random_relation]

  message = f"Hey {nicknames[num1]} I hear that your {random_relationship} named {random_relation},"
  print(message)
```

```python
# We are getting close! What else do we need?
#" Hey {choose a random nickname} I hear that your {choose a random child/pet},
# named {the random child/pets name} also likes to eat {chose a random favorite food}!"

# We need to list a food

for name in my_people.keys():
  nicknames = my_people[name]['Nicknames']
  num_nicknames = len(nicknames)
  num1 = np.random.randint(num_nicknames)

  relations = list(my_people[name]['Family'].keys())
  num_relations = len(relations)
  num2 = np.random.randint(num_relations)
  random_relation = relations[num2]
  random_relationship = my_people[name]['Family'][random_relation]

  random_food = 'food' # Just a place holder

  message = f"Hey {nicknames[num1]} I hear that your {random_relationship} named {random_relation} "
  message += f'also likes to eat {random_food}'
  print(message)
```
