### Doing things with Dictionaries

    - Add data
    - Empty dictionary
    - Modifying data
    - Removing data

```python
cat1 = {
    'name':'Bongo',
    'color':'black',
    'weight':'15'
}
print(cat1)
```

```python
# Lets add some information
cat1['texture']='very soft'
cat1['favorite food'] = 'pill pockets'
print(cat1)
```

```python
# Lets define a new cat and add pieces one by one
cat2 = {} # This is an empty dictioary
cat2['name'] = 'Bella'
cat2['color'] = 'black-brown'
cat2['weight'] = 6
cat2['texture'] = 'soft'
cat2['favorite food'] = 'ice cream'
print(cat2)
```

```python
# What if cat1 gained weight
print(f"{cat1['name']} use to weigh {cat1['weight']} pounds...")
# We would need to modify the data in the weight category
cat1['weight']=16
print(f"{cat1['name']} has been eating too many {cat1['favorite food']} and now weighs {cat1['weight']} pounds!")
```

```python
# Maybe we don't care to keep track of texture
del cat1['texture']
del cat2['texture']
print(cat1,'\n',cat2)
```

### Looping through Dictionaries:

There are some really nice functions for looping through dictionaries:
   - .items()
   - .keys()
   - .values()

```python
my_meals = {
    "breakfast":"cereal, milk, and coffee.",
    "lunch":"spiced harissa rice, carrots, and hummus.",
    'dinner':"pork souvlaki rice bowl with kimchi."
}
```

```python
# Print out information about each of the meals I ate today:

for k,v in my_meals.items():
    print(f"For {k} I ate {v}.")
```

```python
# Just check what meals I ate

print("Today I ate:")
for k in my_meals.keys():
    print(k)
```

```python
# Just check what food I ate:

print('Today I ate:')
for v in my_meals.values():
    print(v.capitalize())
```

```python
# Check to see if I had a snack

snack_info = 'snack' in my_meals.keys()
print(f'I had a snack: {snack_info}')
```

### NESTING

It's not just for the birds!

You can make a list of dictionaries, a dictionary of lists, or a dictionary of dictionaries. It all depends on the form of your data.

It really pays to think a little bit about the best form for your data before launching into a big programming project!

```python
# A list of dictionaries:

person1 = {
        'first':'Joanna',
        'last':'Bieri',
        'age':'44'
}
person2 = {
        'first':'Kristin',
        'last':'Williams',
        'age':'42'
}
person3 = {
        'first':'Stefanie',
        'last':'Bieri',
        'age':'75'
}

people = [person1, person2, person3]

for person in people:
    print(f"{person['first']} {person['last']} is {person['age']} years old")
```

```python
# A dictionary of lists
my_meals = {
    "breakfast":["cereal"," milk"," coffee"],
    "lunch":["spiced harissa rice", "carrots"," hummus"],
    'dinner':["pork souvlaki rice bowl", "kimchi"]
}

for k in my_meals.keys():
    print(f"For {k} I ate {my_meals[k]}")
```

```python
# A dictionary of dictionaries
people = {
    "person1":{
        'first':'Joanna',
        'last':'Bieri',
        'age':'44'
    },
    "person2":{
        'first':'Kristin',
        'last':'Williams',
        'age':'42'
    },
    "person3":{
        'first':'Stefanie',
        'last':'Bieri',
        'age':'75'
    }
}

for person in people.keys():
    print(f"{people[person]['first']} {people[person]['last']} is {people[person]['age']} years old")
```
